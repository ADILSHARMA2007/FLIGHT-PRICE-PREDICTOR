```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

df = pd.read_csv("D:\FLIGHT PRICE PREDICTOR\BASE\indian\Clean_Dataset.csv")

print(df.shape)
print(df.columns.tolist())
df.head()
```

```python
print(df.info())
print("\nNull values:\n", df.isnull().sum())
```

```python
# Price distribution
plt.figure(figsize=(10, 5))
sns.histplot(df["price"], bins=60, kde=True, color="teal")
plt.title("Distribution of Flight Price")
plt.xlabel("Price")
plt.ylabel("Count")
plt.show()
```

```python
# Boxplot for outliers in price
plt.figure(figsize=(10, 3))
sns.boxplot(x=df["price"], color="orange")
plt.title("Price Boxplot")
plt.xlabel("Price")
plt.show()
```

```python
# Flights count by airline
plt.figure(figsize=(12, 6))
airline_counts = df["airline"].value_counts()
sns.barplot(x=airline_counts.index, y=airline_counts.values, palette="viridis")
plt.title("Number of Flights by Airline")
plt.xlabel("Airline")
plt.ylabel("Flight Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# Average price by airline
plt.figure(figsize=(12, 6))
avg_price_airline = df.groupby("airline", as_index=False)["price"].mean().sort_values("price", ascending=False)
sns.barplot(data=avg_price_airline, x="airline", y="price", palette="magma")
plt.title("Average Ticket Price by Airline")
plt.xlabel("Airline")
plt.ylabel("Average Price")
plt.xticks(rotation=45)
plt.show()
```

```python
# Source city distribution
plt.figure(figsize=(10, 5))
source_counts = df["source_city"].value_counts()
sns.barplot(x=source_counts.index, y=source_counts.values, palette="crest")
plt.title("Flights by Source City")
plt.xlabel("Source City")
plt.ylabel("Flight Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# Destination city distribution
plt.figure(figsize=(10, 5))
dest_counts = df["destination_city"].value_counts()
sns.barplot(x=dest_counts.index, y=dest_counts.values, palette="flare")
plt.title("Flights by Destination City")
plt.xlabel("Destination City")
plt.ylabel("Flight Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# Average price by class
plt.figure(figsize=(6, 4))
sns.barplot(data=df, x="class", y="price", estimator=np.mean, ci=None, palette="Set2")
plt.title("Average Price by Class")
plt.xlabel("Class")
plt.ylabel("Average Price")
plt.show()
```

```python
# Average price by number of stops
plt.figure(figsize=(7, 5))
order_stops = ["zero", "one", "two_or_more"]
sns.barplot(data=df, x="stops", y="price", order=order_stops, estimator=np.mean, ci=None, palette="coolwarm")
plt.title("Average Price by Stops")
plt.xlabel("Stops")
plt.ylabel("Average Price")
plt.show()
```

```python
# Duration vs price
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df.sample(min(20000, len(df)), random_state=42), x="duration", y="price", alpha=0.35)
plt.title("Duration vs Price")
plt.xlabel("Duration (hours)")
plt.ylabel("Price")
plt.show()
```

```python
# Days left vs price
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df.sample(min(20000, len(df)), random_state=42), x="days_left", y="price", alpha=0.35, color="purple")
plt.title("Days Left vs Price")
plt.xlabel("Days Left")
plt.ylabel("Price")
plt.show()
```

```python
# Heatmap of average price: source -> destination
pivot_route = df.pivot_table(index="source_city", columns="destination_city", values="price", aggfunc="median")

plt.figure(figsize=(10, 7))
sns.heatmap(pivot_route, annot=True, fmt=".0f", cmap="YlOrRd")
plt.title("Average Price Heatmap (Source vs Destination)")
plt.xlabel("Destination City")
plt.ylabel("Source City")
plt.show()
```

```python
# Departure time vs average price (with class split)
plt.figure(figsize=(12, 6))
dep_order = ["Early_Morning", "Morning", "Afternoon", "Evening", "Night", "Late_Night"]
sns.barplot(data=df, x="departure_time", y="price", hue="class", order=dep_order, estimator=np.mean, ci=None)
plt.title("Average Price by Departure Time and Class")
plt.xlabel("Departure Time")
plt.ylabel("Average Price")
plt.xticks(rotation=45)
plt.legend(title="Class")
plt.show()
```

```python
# Arrival time vs average price
plt.figure(figsize=(12, 6))
arr_order = ["Early_Morning", "Morning", "Afternoon", "Evening", "Night", "Late_Night"]
sns.barplot(data=df, x="arrival_time", y="price", order=arr_order, estimator=np.mean, ci=None, palette="Spectral")
plt.title("Average Price by Arrival Time")
plt.xlabel("Arrival Time")
plt.ylabel("Average Price")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sample_df = df.sample(min(30000, len(df)), random_state=42)

sns.regplot(
    data=sample_df,
    x="duration",
    y="price",
    scatter_kws={"alpha": 0.25, "s": 12},
    line_kws={"color": "red", "linewidth": 2}
)

plt.title("Regression Line: Duration vs Price")
plt.xlabel("Duration (hours)")
plt.ylabel("Price")
plt.show()
```

```python
#DAY LEFT VS PRICE
plt.figure(figsize=(10, 6))
sample_df = df.sample(min(30000, len(df)), random_state=42)

sns.regplot(
    data=sample_df,
    x="days_left",
    y="price",
    scatter_kws={"alpha": 0.25, "s": 12},
    line_kws={"color": "darkgreen", "linewidth": 2}
)

plt.title("Regression Line: Days Left vs Price")
plt.xlabel("Days Left")
plt.ylabel("Price")
plt.show()
```

```python
# Top 20 most expensive flights by flight code (average)
top_flights = (
    df.groupby("flight", as_index=False)["price"]
    .mean()
    .sort_values("price", ascending=False)
    .head(20)
)

plt.figure(figsize=(12, 6))
sns.barplot(data=top_flights, x="flight", y="price", palette="rocket")
plt.title("Top 20 Flight Codes by Average Price")
plt.xlabel("Flight Code")
plt.ylabel("Average Price")
plt.xticks(rotation=60)
plt.show()
```

```python
# Correlation heatmap (numeric columns)
num_df = df.select_dtypes(include=[np.number])

plt.figure(figsize=(6, 4))
sns.heatmap(num_df.corr(), annot=True, cmap="Blues", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Violin plot: price distribution by stops
plt.figure(figsize=(8, 5))
order_stops = ["zero", "one", "two_or_more"]
sns.violinplot(data=df, x="stops", y="price", order=order_stops, palette="Pastel1")
plt.title("Price Distribution by Stops")
plt.xlabel("Stops")
plt.ylabel("Price")
plt.show()
```
